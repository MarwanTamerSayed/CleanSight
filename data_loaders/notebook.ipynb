```python
from langchain_core.documents import Document
```

```python
def pdf_loader(folder_path):
    documnets = []
    for file in folder_path.glob("*.pdf"):
        doc = fitz.open(file)
        
        text = ""
        text += f"\n--- {file.name} ---"
        for page in doc:
            text += page.get_text()

        documnet = Document(page_content=text,metadata={'source':file})
        documnets.append(documnet)

    return documnets    
```

```python
from langchain_core.documents import Document
```

```python
def csv_loader(folder_path):
    folder_path = Path(folder_path)
    documents = []
    for file in folder_path.glob('*.csv'):
        df = pd.read_csv(file)

        for _, row in df.iterrows():
            text = "\n".join(
                [f"{col}: {value}" for col, value in row.items()]
            )

            documents.append(
                Document(
                    page_content=text,
                    metadata={"source": file}
                )
            )

    return documents        
            
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
```

```python
len(embeddings.embed_query("how are u today ?"))
```

```python
import os
from langchain_community.vectorstores import Chroma

BASE_DB_DIR = os.path.abspath("../chroma_dbs")

def create_vector_store(chunks, db_name="default_db"):
    db_path = os.path.join(BASE_DB_DIR, db_name)

    # Create folder if it doesn't exist
    os.makedirs(db_path, exist_ok=True)

    # Check if Chroma DB already exists
    if os.listdir(db_path):
        print(f"Loading existing DB: {db_name}")

        vector_store = Chroma(
            persist_directory=db_path,
            embedding_function=embeddings
        )

        # Add new documents
        vector_store.add_documents(chunks)

    else:
        print(f"Creating new DB: {db_name}")

        vector_store = Chroma.from_documents(
            documents=chunks,
            embedding=embeddings,
            persist_directory=db_path
        )

    vector_store.persist()

    return vector_store
```

```python
db=create_vector_store(chunks, db_name="pdf_reports_db")
```
